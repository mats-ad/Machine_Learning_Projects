# Same ML Problem with and without Feature Engineering

Dataset: California Housing

Model: Linear Regression

Linear Regression is interesting because when Feature Engineering works on a simple Model like LR you could transfer it to mor complex models

Metric: Root Mean Squared Error (RMSE)

```python
data
```

```python
df = data.frame
df.head()
```

```python
X = df.drop("MedHouseVal", axis=1)
X.head()
```

```python
y = df["MedHouseVal"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

```python
baseline_pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("model", LinearRegression())
])

baseline_pipeline.fit(X_train, y_train)
```

```python
y_pred_baseline = baseline_pipeline.predict(X_test)
```

```python
rmse_baseline = root_mean_squared_error(y_test, y_pred_baseline)
rmse_baseline
```

```python
X
```

```python
X_fe = X.copy()

X_fe["BedroomsPerRoom"] = X_fe["AveBedrms"] / X_fe["AveRooms"]
```

```python
X_fe["PopulationDividedByMedInc"] = X_fe["Population"] / X_fe["MedInc"]
```

```python
X_train_fe, X_test_fe, y_train_fe, y_test_fe = train_test_split(X_fe, y, test_size=0.2)
```

```python
fe_pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("model", LinearRegression())
])

fe_pipeline.fit(X_train_fe, y_train_fe)
```

```python
y_pred_fe = fe_pipeline.predict(X_test_fe)
```

```python
rmse_fe = root_mean_squared_error(y_test_fe, y_pred_fe)
rmse_fe
```

```python
print(rmse_baseline)
print(rmse_fe)
```

As we can see the through minimal Feature Engineering the models performance improved when exposing meaningful relationships
